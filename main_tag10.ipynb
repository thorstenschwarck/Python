```python
#### 
# Baue ein Ratespiel: Das Programm denkt sich zu Beginn eine zufällige ganze Zahl aus einem Zahlenbereich deiner Wahl aus. 
# Danach darf der Spieler so lange raten, bis er die Zahl gefunden hat. 
# Nach jedem Tipp verrät das Programm, ob die gesuchte Zahl größer oder kleiner ist. 
# Gibt der Spieler etwas ein, das gar keine ganze Zahl ist, stürzt das Programm nicht ab, 
# sondern bittet freundlich um eine neue Eingabe. 
# Liegt die eingegebene Zahl außerhalb deines Zahlenbereichs, gibt es ebenfalls nur einen Hinweis und es geht weiter. 
# Wird die Zahl erraten, gratuliert das Programm und verrät, wie viele gültige Versuche gebraucht wurden.

####
```

```python
import random
```

```python
geheimzahl = random.randint(1, 100)
tipp = 0

while tipp != geheimzahl:
    try:
        tipp = int(input("Dein Tipp(Ganze Zahl zwischen 1 und 100): "))
    except ValueError:
        print("Bitte eine ganze Zahl eingeben!")    
        continue
    if tipp < geheimzahl:
        print("Tipp ist zu niedrig!")
    elif tipp > geheimzahl:
        print("Dein Tipp ist zu hoch!")

print("Richtig! Geheimzahl der Maschine war: {geheimzahl}")
    
```
